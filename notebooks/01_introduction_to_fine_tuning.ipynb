# 01 — Introduction to Fine-Tuning

Welcome to the **"Hello World" of fine-tuning**! 🎉

## What is fine-tuning?

Large Language Models (LLMs) are pre-trained on massive amounts of text. They learn *language* — but not necessarily *your task*.

**Fine-tuning** = taking a pre-trained model and continuing training it on a smaller, task-specific dataset.

| Analogy | Pre-training | Fine-tuning |
|---------|--------------|-------------|
| School | 12 years of general education | A 3-month course in one skill |
| What it learns | Grammar, facts, reasoning | Your specific task/style |
| Data size | Trillions of tokens | Hundreds to millions of examples |
| Cost | Millions of dollars | A few GPU-hours (or free on Colab!) |

## Why fine-tune?

1. **Better performance on a specific task** (classification, style, format)
2. **Smaller model, same quality** — fine-tuned 0.5B can beat prompted 7B
3. **Cost savings** — fewer tokens in prompts, no few-shot examples needed

## When NOT to fine-tune?

- Your task can be solved with good **prompting** or **RAG** (retrieval)
- You have **very little data** (< ~50 examples)
- You just need **knowledge** injection → use RAG instead

> Rule of thumb: **prompting changes behavior a little, fine-tuning changes behavior a lot, RAG changes what the model knows.**

## Types of fine-tuning

### 1. Supervised Fine-Tuning (SFT) — *what we do in this repo*
Train on `(instruction → response)` pairs. The model learns to follow your format/task.

### 2. Continued Pre-training
Train on raw domain text (e.g., legal documents). The model absorbs the domain.

### 3. RLHF / DPO
Align with human preferences using preference data (chosen vs rejected responses).

## The data format (Alpaca style)

Most SFT datasets look like this:

```json
{
  "instruction": "Translate to formal English.",
  "input": "hey, can u help me?",
  "output": "Hello, would you kindly assist me?"
}
```

Let's look at the tiny dataset we'll use in this project:

````python
import json

with open('../data/hello_alpaca.json') as f:
    dataset = json.load(f)

print(f"Number of examples: {len(dataset)}")
print(json.dumps(dataset[2], indent=2, ensure_ascii=False))
````

````python
# How a chat model actually sees this data (simplified):

example = dataset[2]
prompt = f"""<|im_start|>user
{example['instruction']}\n{example['input']}<|im_end|>
<|im_start|>assistant
{example['output']}<|im_end|>"""

print(prompt)
print("\n--- Only the 'assistant' part contributes to the loss during training!")
````

## Summary

- Fine-tuning = continuing training on task-specific data
- SFT on instruction data is the most common and approachable form
- 50–100 good examples can already teach a style or format
